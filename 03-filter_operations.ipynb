```python
df = spark.read.option('header', 'true').csv('BigMart_Sales.csv')
```

```python
df.show(3)
```

### 1. Filter Operations

```python
from pyspark.sql.functions import col

df.filter(col("Item_Type") == "Dairy").show(5)
```

```python
df.filter((col("Item_Type") == "Dairy")).show(5)
```

#### 1.1 Multiple Filter Conditions

```python
df.filter((col("Item_Type") == "Dairy") & (col("Item_Fat_Content") == "Regular")).show(5)
```

- Not equal to given condition

```python
df.filter(~(col("Item_Type") == "Dairy")).show(5)
```
