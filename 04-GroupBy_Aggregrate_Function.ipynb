### PySpark GroupBy And Aggregate Function

```python
# Using SparkSession
spark.sparkContext.setLogLevel("ERROR")
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.types import IntegerType, DoubleType
```

```python
spark = SparkSession.builder.appName('practice').getOrCreate()
```

```python
df = spark.read.option('header', 'true').csv('salary.csv')
```

```python
df.show(3)
```

```python
df.describe()
```

```python
df.columns
```

```python
df.printSchema()
```

```python
df = df.withColumn('salary', df['salary'].cast(DoubleType()))
```

### 1. GroupBy

```python
df.groupBy('name').sum().show()
```

```python
df.groupBy('department').sum().show()
```

```python
df.groupBy('department').mean().show()
```

```python
df.groupBy('department').count().show()
```

```python
df.agg({'Salary':'Sum'}).show()
```

```python
df.groupBy('department').max().show()
```

```python
df.groupBy('department').avg().show()
```
