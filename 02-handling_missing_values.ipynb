```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('practice').getOrCreate()
```

```python
spark
```

```python
df = spark.read.option('header', 'true').csv('BigMart_Sales.csv')
```

```python
df.show(3)
```

### 1. Dropping Rows

```python
df = df.na.drop()
```

```python
df.show()
```

#### 1.1 Any/All

Drops rows even if any of values is null

```python
df = df.na.drop(how='any')
```

Drops rows only when all the values are null

```python
df = df.na.drop(how=all)
```

#### 1.2 Threshold

- Threshold says atleast mention number of non nulls should be present
- Here atleast 2 non null values should be present

```python
df = df.na.drop(how='any', thresh=2)
```

#### 1.3 Subset

Using Subset we can remove all the null values from specified column subset

```python
df = df.na.drop(how='any', subset=['Item_Type'])
```

### 2. Filling Missing Values

```python
df = df.na.fill('Missing Value')
```

```python
df.show(5)
```

We can do this for specific columns as well by specifying column name/names 

```python
df = df.na.fill('Missing Value', 'Outlet_Identifier')
```

```python
df = df.na.fill('Missing Value', ['Outlet_Identifier', 'Outlet_Establishment_Year'])
```

#### Using Imputer Method

```python
from pyspark.ml.feature import Imputer

imputer = Imputer(
    inputCol=['Item_Weight', 'Item_MRP'],
    outputCol=["{}_imputed".format(c) for c in ['Outlet_Identifier', 'Outlet_Establishment_Year']]
).setStrategy('mean')
```

Adding imputation cols to DF

```python
imputer.fit(df).transform(df).show()
```
